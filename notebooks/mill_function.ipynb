```python
import numpy as np
import pandas as pd

def fit_one(table, feature):
    a, b = np.polyfit(table[feature], table["VB"], 1)
    VB_estimated = a*table[feature] + b
    VB_residual = table["VB"] - VB_estimated
    VB_result = pd.DataFrame({
        "run number":table["run number"],
        "VB": table["VB"],
        "estimated VB":VB_estimated,
        "VB residual":VB_residual
    })
    return a,b,VB_result


def fit_two(table, feature1, feature2):
    X = np.column_stack((
        table[feature1],
        table[feature2],
        np.ones(len(table))
    ))
    a, b, c = np.linalg.lstsq(X, table["VB"], rcond=None)[0]
    VB_estimated = a*table[feature1] + b*table[feature2] + c
    VB_residual = table["VB"] - VB_estimated
    VB_result = pd.DataFrame({
        "run number":table["run number"],
        "VB": table["VB"],
        "estimated VB":VB_estimated,
        "VB residual":VB_residual
    })
    return a, b, c,VB_result
```
